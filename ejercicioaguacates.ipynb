Ejercicio: "Concurso de Aguacates Gigantes"

Un grupo de agricultores está ahora compitiendo para ver quién tiene los aguacates más pesados.

 Esta vez, además de analizar los datos básicos (media, mediana, moda, desviación estándar), necesitarán también:



Calcular:


Media (promedio) del peso.


Mediana.


Moda.


Desviación estándar.


Coeficiente de variación (desviación estándar / media * 100) → Para ver qué tan homogéneo es el cultivo.
Identificar:



El aguacate más pesado y el más ligero.

Cuántos aguacates están dentro de un rango "ideal" de peso (por ejemplo, entre 400g y 600g).

Visualizar:



Un histograma de distribución de pesos.

Una gráfica de caja ("boxplot") para visualizar los valores atípicos (outliers).
Interactividad:

Permitir al usuario definir el rango ideal (no fijo como 400-600g).

Permitir ingresar datos manualmente o usar un conjunto de datos precargados.
Extras Nivel Avanzado  (opcional pero recomendado):

Calcular la asimetría del conjunto de datos (indicador de qué tan simétrica es la distribución).

Guardar el análisis completo en un archivo .txt o .csv.

```python
import statistics
import matplotlib.pyplot as plt

# Elegir los datos
opcion = input("¿Quieres datos precargados (1) o introducirlos manualmente (2)? ")

if opcion == "1":
    aguacates = [450, 520, 600, 480, 550, 700, 390, 520, 580, 450]

else:
    aguacates = []
    cantidad = int(input("¿Cuántos aguacates vas a introducir? "))

    for i in range(cantidad):
        peso = float(input("Introduce el peso en gramos: "))
        aguacates.append(peso)

# Definir el rango ideal
peso_minimo = float(input("Peso mínimo ideal: "))
peso_maximo = float(input("Peso máximo ideal: "))

# Calcular estadísticas
media = statistics.mean(aguacates)
mediana = statistics.median(aguacates)
moda = statistics.mode(aguacates)
desviacion = statistics.pstdev(aguacates)

# Coeficiente de variación
coeficiente = desviacion / media * 100

# Calcular la asimetría
media_cubica = sum((peso - media) ** 3 for peso in aguacates) / len(aguacates)
asimetria = media_cubica / (desviacion ** 3) if desviacion != 0 else 0

# Encontrar el peso máximo y mínimo
mas_pesado = max(aguacates)
mas_ligero = min(aguacates)

# Contar los aguacates ideales
contador = 0

for peso in aguacates:
    if peso_minimo <= peso <= peso_maximo:
        contador = contador + 1

# Mostrar resultados
print("\n--- RESULTADOS DEL CONCURSO ---")
print("Número de aguacates:", len(aguacates))
print("Media:", round(media, 2), "gramos")
print("Mediana:", round(mediana, 2), "gramos")
print("Moda:", moda, "gramos")
print("Desviación estándar:", round(desviacion, 2))
print("Coeficiente de variación:", round(coeficiente, 2), "%")
print("Asimetría:", round(asimetria, 3))
print("Más pesado:", mas_pesado, "gramos")
print("Más ligero:", mas_ligero, "gramos")
print("Dentro del rango ideal:", contador)

# Crear las gráficas
plt.hist(aguacates, bins=5, edgecolor="black")
plt.title("Distribución de pesos")
plt.xlabel("Peso en gramos")
plt.ylabel("Frecuencia")
plt.show()

plt.boxplot(aguacates)
plt.title("Gráfica de caja")
plt.ylabel("Peso en gramos")
plt.show()

# Guardar los resultados en un archivo TXT
with open("analisis_aguacates.txt", "w", encoding="utf-8") as archivo:
    archivo.write("--- RESULTADOS DEL CONCURSO ---\n")
    archivo.write(f"Número de aguacates: {len(aguacates)}\n")
    archivo.write(f"Media: {media:.2f} gramos\n")
    archivo.write(f"Mediana: {mediana:.2f} gramos\n")
    archivo.write(f"Moda: {moda} gramos\n")
    archivo.write(f"Desviación estándar: {desviacion:.2f}\n")
    archivo.write(f"Coeficiente de variación: {coeficiente:.2f}%\n")
    archivo.write(f"Asimetría: {asimetria:.3f}\n")
    archivo.write(f"Aguacate más pesado: {mas_pesado} gramos\n")
    archivo.write(f"Aguacate más ligero: {mas_ligero} gramos\n")
    archivo.write(f"Dentro del rango ideal: {contador}\n")

print("\nAnálisis guardado en analisis_aguacates.txt")
```
